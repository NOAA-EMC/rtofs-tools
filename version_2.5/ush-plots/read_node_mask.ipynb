```python
import os

import numpy as np

import matplotlib.pyplot as plt

%matplotlib inline
```

```python
def read_node_mask(file_path, cols=4500, rows=3298):
    # Dimensions based on your 1o4500x3298 filenames
    
    if not os.path.exists(file_path):
        print(f"Error: {file_path} not found.")
        return None

    try:
        # 1. Read raw 4-byte data (matches Fortran access='stream')
        # Fortran code is likely compiled on a system that defaults to 
        # Big Endian for unformatted files, or the file was generated on such a system.

        # '>i4' means:
        # '>' = Big Endian
        # 'i' = integer
        # '4' = 4 bytes
        data = np.fromfile(file_path, dtype='>i4')
        node_mask = data.reshape((rows, cols), order='F')
        
        # 2. Reshape using 'F' (Fortran/Column-major) order
        # This is CRITICAL for binary files created in Fortran
        node_msk = data.reshape((rows, cols), order='F')
        
        print(f"Successfully read: {file_path}")
        print(f"Shape: {node_msk.shape}")
        return node_msk
        
    except Exception as e:
        print(f"An error occurred: {e}")
        return None
```

# set path to `codaclim` file(s) and files to read

```python
codaclim_path = "/scratch5/NCEPDEV/rstprod/Santha.Akella/data/rtofs/v2p5/ncoda/codaclim/"
file_name = codaclim_path + "HYCOM.node_mask"
```

```python
path = "/scratch5/NCEPDEV/rstprod/Santha.Akella/data/rtofs/v2p5/ncoda/codaclim/HYCOM.node_mask"
mask = read_node_mask(file_name)

if mask is not None:
    print("Top-left 5x5 corner of mask:")
    print(mask[:5, :5])
```

```python
# 2. Optimized Plotting
plt.figure(figsize=(6, 4))

img = plt.imshow(mask, 
                 cmap='gray', 
                 origin='lower', 
                 interpolation='nearest',
                 aspect='auto')

plt.colorbar(img, ticks=[0, 1], label='0: Land | 1: Sea')
plt.title(f"HYCOM Node Mask ({mask.shape[1]}x{mask.shape[0]})", fontsize=14)
plt.xlabel("Longitude Index")
plt.ylabel("Latitude Index")

plt.tight_layout()
plt.show()
```
